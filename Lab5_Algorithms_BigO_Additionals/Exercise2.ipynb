```python
# [CELL 1] STUDENT CONFIGURATION
LAST_NAME = 'Magalong'
STUDENT_ID = 'TUPM-26-1134'
FAVORITE_ARTIST = 'Rico Blanco'

SEED_DIGIT = int(STUDENT_ID[-1])
ID_SUM = sum(int(d) for d in STUDENT_ID if d.isdigit())
NAME_LENGTH = len(LAST_NAME)

letters = [c for c in FAVORITE_ARTIST.upper() if c.isalpha()]
ARTIST_LENGTH = len(letters)
ARTIST_SUM = sum(ord(c) - ord('A') + 1 for c in letters)

print('Last Name     :', LAST_NAME)
print('Student ID    :', STUDENT_ID)
print('Fav. Artist   :', FAVORITE_ARTIST)
print('SEED_DIGIT    :', SEED_DIGIT)
print('ID_SUM        :', ID_SUM)
print('NAME_LENGTH   :', NAME_LENGTH)
print('ARTIST_LENGTH :', ARTIST_LENGTH)
print('ARTIST_SUM    :', ARTIST_SUM)
```

```python
# [CELL 2] DATA GENERATION
size_a = 8 + SEED_DIGIT
size_b = NAME_LENGTH + 2

dataset_a = [SEED_DIGIT * (i + 1) + ID_SUM * i for i in range(size_a)]
dataset_b = [ARTIST_LENGTH * (i + 1) + NAME_LENGTH * i for i in range(size_b)]

print('Department A:', dataset_a)
print('Department B:', dataset_b)
print('Both ascending:', dataset_a == sorted(dataset_a) and dataset_b == sorted(dataset_b))
```

```python
# [CELL 3] MERGE ALGORITHM (two-pointer)
def merge(a, b):
    merged, i, j, comparisons = [], 0, 0, 0
    while i < len(a) and j < len(b):
        comparisons += 1
        if a[i] <= b[j]:
            merged.append(a[i]); i += 1
        else:
            merged.append(b[j]); j += 1
    merged.extend(a[i:])
    merged.extend(b[j:])
    return merged, comparisons

consolidated, comparisons = merge(dataset_a, dataset_b)

print('Consolidated:', consolidated)
print('Comparisons :', comparisons)
```

```python
# [CELL 4] VERIFICATION
before = len(dataset_a) + len(dataset_b)
after = len(consolidated)
is_ordered = all(consolidated[k] <= consolidated[k + 1] for k in range(len(consolidated) - 1))
matches_builtin = consolidated == sorted(dataset_a + dataset_b)

print('Records before consolidation:', len(dataset_a), '+', len(dataset_b), '=', before)
print('Records after consolidation :', after)
print('Count preserved             :', before == after)
print('Ordering verified           :', is_ordered)
print('Matches sorted(A + B)       :', matches_builtin)
print('Comparison bounds           :', f'min(m,n)={min(size_a, size_b)} <= {comparisons} <= m+n-1={before - 1}')
```

```python
# [CELL 5] COMPLEXITY SUMMARY
m, n_ = len(dataset_a), len(dataset_b)
print(f'm = {m}, n = {n_}')
print('Time Complexity  : O(m + n)')
print('Space Complexity : O(m + n)  (new merged list)')
```

```python
# [CELL 6] FINAL OUTPUT
print('=' * 50)
print('FINAL OUTPUT - UNIVERSITY ENROLLMENT RECORDS')
print('=' * 50)
print('Department A dataset :', dataset_a)
print('Department B dataset :', dataset_b)
print('Consolidated dataset :', consolidated)
print('Records (before/after):', before, '/', after)
print('Comparison count     :', comparisons)
print('Ordering verification:', 'PASSED' if is_ordered and before == after else 'FAILED')
print('Time / Space         : O(m + n) / O(m + n)')
```
