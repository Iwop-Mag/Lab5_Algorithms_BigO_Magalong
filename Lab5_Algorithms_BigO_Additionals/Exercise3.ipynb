```python
# [CELL 1] STUDENT CONFIGURATION
LAST_NAME = 'Magalong'
STUDENT_ID = 'TUPM-26-1134'
FAVORITE_ARTIST = 'Rico Blanco'

SEED_DIGIT = int(STUDENT_ID[-1])
ID_SUM = sum(int(d) for d in STUDENT_ID if d.isdigit())
NAME_LENGTH = len(LAST_NAME)

letters = [c for c in FAVORITE_ARTIST.upper() if c.isalpha()]
ARTIST_LENGTH = len(letters)
ARTIST_SUM = sum(ord(c) - ord('A') + 1 for c in letters)

print('Last Name     :', LAST_NAME)
print('Student ID    :', STUDENT_ID)
print('Fav. Artist   :', FAVORITE_ARTIST)
print('SEED_DIGIT    :', SEED_DIGIT)
print('ID_SUM        :', ID_SUM)
print('NAME_LENGTH   :', NAME_LENGTH)
print('ARTIST_LENGTH :', ARTIST_LENGTH)
print('ARTIST_SUM    :', ARTIST_SUM)
```

```python
# [CELL 2] ORDER GENERATION
import random

n_orders = 100 + ID_SUM * 5
order_numbers = [1000 + i * (SEED_DIGIT + 1) + (i % NAME_LENGTH) for i in range(n_orders)]
order_numbers = sorted(set(order_numbers))          # unique order numbers

sorted_input = order_numbers.copy()
reverse_input = order_numbers[::-1]
random_input = order_numbers.copy()
random.Random(ID_SUM * SEED_DIGIT + ARTIST_LENGTH).shuffle(random_input)   # reproducible shuffle

print('Number of orders :', len(order_numbers))
print('Sorted  (first 8):', sorted_input[:8])
print('Reverse (first 8):', reverse_input[:8])
print('Random  (first 8):', random_input[:8])
print('Same set of orders:', set(sorted_input) == set(reverse_input) == set(random_input))
```

```python
# [CELL 3] SORTING ALGORITHMS
def bubble_sort(data):
    a, comparisons, swaps = data.copy(), 0, 0
    for i in range(len(a)):
        swapped = False
        for j in range(len(a) - i - 1):
            comparisons += 1
            if a[j] > a[j + 1]:
                a[j], a[j + 1] = a[j + 1], a[j]
                swaps += 1
                swapped = True
        if not swapped:
            break
    return a, comparisons, swaps


def selection_sort(data):
    a, comparisons, swaps = data.copy(), 0, 0
    for i in range(len(a) - 1):
        minimum = i
        for j in range(i + 1, len(a)):
            comparisons += 1
            if a[j] < a[minimum]:
                minimum = j
        if minimum != i:
            a[i], a[minimum] = a[minimum], a[i]
            swaps += 1
    return a, comparisons, swaps


def insertion_sort(data):
    a, comparisons, shifts = data.copy(), 0, 0
    for i in range(1, len(a)):
        key, j = a[i], i - 1
        while j >= 0:
            comparisons += 1
            if a[j] <= key:
                break
            a[j + 1] = a[j]
            shifts += 1
            j -= 1
        a[j + 1] = key
    return a, comparisons, shifts
```

```python
# [CELL 4] RUN ALL ALGORITHMS ON ALL CONDITIONS
import time

def measure(sort_function, data, trials=5):
    times = []
    for _ in range(trials):
        start = time.perf_counter()
        result = sort_function(data)
        times.append(time.perf_counter() - start)
    return result, sum(times) / trials

conditions = {'Sorted': sorted_input, 'Reverse': reverse_input, 'Random': random_input}
algorithms = {'Bubble Sort': bubble_sort, 'Selection Sort': selection_sort, 'Insertion Sort': insertion_sort}
expected = sorted(order_numbers)

records = []
for cond, data in conditions.items():
    for name, fn in algorithms.items():
        (result, comps, moves), elapsed = measure(fn, data)
        records.append((cond, name, comps, moves, elapsed, result == expected))

print(f'{"Condition":<10}{"Algorithm":<16}{"Comparisons":>12}{"Movements":>11}{"Time (s)":>13}{"Sorted?":>9}')
for cond, name, comps, moves, elapsed, ok in records:
    print(f'{cond:<10}{name:<16}{comps:>12}{moves:>11}{elapsed:>13.8f}{str(ok):>9}')
```

```python
# [CELL 5] SORTING VERIFICATION
print('All 9 runs correctly sorted:', all(r[5] for r in records))
```

```python
# [CELL 6] CASE OBSERVATIONS (from measured data)
N = len(order_numbers)
print(f'n = {N}  |  n(n-1)/2 = {N*(N-1)//2}  |  n-1 = {N-1}\n')
for name in algorithms:
    rows = {r[0]: r for r in records if r[1] == name}
    fastest = min(rows.values(), key=lambda r: r[2])[0]
    slowest = max(rows.values(), key=lambda r: r[2])[0]
    print(f'{name}:')
    for cond in conditions:
        print(f'   {cond:<8} comparisons={rows[cond][2]:>6}  movements={rows[cond][3]:>6}')
    print(f'   fewest comparisons on: {fastest} | most on: {slowest}\n')
```

```python
# [CELL 7] BIG O SUMMARY
print(f'{"Algorithm":<16}{"Best":<10}{"Average":<10}{"Worst":<10}{"Space":<8}')
print(f'{"Bubble Sort":<16}{"O(n)":<10}{"O(n^2)":<10}{"O(n^2)":<10}{"O(1)":<8}')
print(f'{"Selection Sort":<16}{"O(n^2)":<10}{"O(n^2)":<10}{"O(n^2)":<10}{"O(1)":<8}')
print(f'{"Insertion Sort":<16}{"O(n)":<10}{"O(n^2)":<10}{"O(n^2)":<10}{"O(1)":<8}')
```

```python
# [CELL 8] FINAL OUTPUT
print('=' * 50)
print('FINAL OUTPUT - E-COMMERCE ORDER PROCESSING')
print('=' * 50)
print('Orders generated :', N)
for cond, name, comps, moves, elapsed, ok in records:
    print(f'{cond:<8}| {name:<15}| comps={comps:>6} | moves={moves:>6} | {elapsed:.8f}s | sorted={ok}')
```
