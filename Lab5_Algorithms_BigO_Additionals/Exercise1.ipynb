```python
# [CELL 1] STUDENT CONFIGURATION
LAST_NAME = 'Magalong'
STUDENT_ID = 'TUPM-26-1134'
FAVORITE_ARTIST = 'Rico Blanco'

SEED_DIGIT = int(STUDENT_ID[-1])
ID_SUM = sum(int(d) for d in STUDENT_ID if d.isdigit())
NAME_LENGTH = len(LAST_NAME)

letters = [c for c in FAVORITE_ARTIST.upper() if c.isalpha()]
ARTIST_LENGTH = len(letters)
ARTIST_SUM = sum(ord(c) - ord('A') + 1 for c in letters)

print('Last Name     :', LAST_NAME)
print('Student ID    :', STUDENT_ID)
print('Fav. Artist   :', FAVORITE_ARTIST)
print('SEED_DIGIT    :', SEED_DIGIT)
print('ID_SUM        :', ID_SUM)
print('NAME_LENGTH   :', NAME_LENGTH)
print('ARTIST_LENGTH :', ARTIST_LENGTH)
print('ARTIST_SUM    :', ARTIST_SUM)
```

```python
# [CELL 2] DATA GENERATION
n = 20 + NAME_LENGTH                       # number of incident records
step = SEED_DIGIT + 1
original = [100 + ID_SUM + i * step for i in range(n)]   # unique, ascending

rotation = SEED_DIGIT                      # rotation point from SEED_DIGIT
rotated = original[rotation:] + original[:rotation]      # first part moved to the end

target = original[(ID_SUM + ARTIST_SUM) % n]             # target from ID_SUM and ARTIST_SUM

print('Records     :', n)
print('Rotation    :', rotation)
print('Original    :', original)
print('Rotated     :', rotated)
print('Target      :', target)
```

```python
# [CELL 3] SEARCH ALGORITHMS
def rotated_binary_search(data, target):
    """Modified binary search for a rotated sorted list. Returns (index, comparisons)."""
    low, high, comparisons = 0, len(data) - 1, 0
    while low <= high:
        mid = (low + high) // 2
        comparisons += 1
        if data[mid] == target:
            return mid, comparisons
        comparisons += 1
        if data[low] <= data[mid]:                 # left half is sorted
            comparisons += 1
            if data[low] <= target < data[mid]:
                high = mid - 1
            else:
                low = mid + 1
        else:                                      # right half is sorted
            comparisons += 1
            if data[mid] < target <= data[high]:
                low = mid + 1
            else:
                high = mid - 1
    return -1, comparisons


def linear_search(data, target):
    for i, value in enumerate(data):
        if value == target:
            return i, i + 1
    return -1, len(data)
```

```python
# [CELL 4] SEARCH RESULTS
bi, bc = rotated_binary_search(rotated, target)
li, lc = linear_search(rotated, target)

print('Target Incident :', target)
print(f'Modified Binary -> index = {bi}, comparisons = {bc}')
print(f'Linear Search   -> index = {li}, comparisons = {lc}')
print('Position match  :', bi == li)

# Not-found demonstration
missing = original[-1] + 1
mi, mc = rotated_binary_search(rotated, missing)
print(f'\nNot-found test ({missing}): index = {mi} ->',
      'NOT FOUND' if mi == -1 else 'found', f'| comparisons = {mc}')
```

```python
# [CELL 5] VERIFY ON EVERY RECORD
ok = all(rotated[rotated_binary_search(rotated, v)[0]] == v for v in rotated)
print('Every record located correctly:', ok)
```

```python
# [CELL 6] COMPLEXITY SUMMARY
print(f'{"Algorithm":<26}{"Time (worst)":<16}{"Space":<8}')
print(f'{"Modified Binary Search":<26}{"O(log n)":<16}{"O(1)":<8}')
print(f'{"Linear Search":<26}{"O(n)":<16}{"O(1)":<8}')
```

```python
# [CELL 7] FINAL OUTPUT
print('=' * 50)
print('FINAL OUTPUT - EMERGENCY DISPATCH RECORD SEARCH')
print('=' * 50)
print('Original Dataset :', original)
print('Rotated Dataset  :', rotated)
print('Target Incident  :', target)
print('Position (Binary):', bi if bi != -1 else 'Not found')
print('Comparisons      : Binary =', bc, '| Linear =', lc)
print('Reduction        :', lc - bc)
print('Time Complexity  : O(log n) vs O(n)')
print('Space Complexity : O(1)')
```
